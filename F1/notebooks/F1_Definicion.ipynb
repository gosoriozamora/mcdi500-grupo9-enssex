## Presentación y contextualización

La Encuesta Nacional de Salud, Sexualidad y Género (ENSSEX) 2022–2023, desarrollada por el Ministerio de Salud de Chile, permite conocer distintas dimensiones de la salud, la sexualidad y el bienestar de la población adulta. Su propósito es generar evidencia que apoye el conocimiento de la salud sexual y reproductiva y el desarrollo de políticas públicas. La encuesta considera a personas de 18 años o más y la base utilizada reúne 20.392 participantes (Ministerio de Salud, s. f.; Ministerio de Salud, 2022).

La amplitud temática de ENSSEX permite estudiar también las relaciones de pareja y las condiciones sociales y económicas. Nuestro proyecto se concentra en las personas encuestadas que conviven actualmente con su pareja y relaciona tres dimensiones: la valoración de la vida sexual, el bienestar mental o emocional percibido y el grado de dependencia económica respecto de la pareja. La base contiene preguntas que permiten identificar la existencia de pareja, la convivencia y la dependencia económica declarada.

Nos interesa conocer si las respuestas sobre estos dos ámbitos del bienestar presentan diferencias entre quienes dependen completamente de su pareja, quienes dependen en parte y quienes declaran no depender económicamente de ella. El análisis será descriptivo y comparativo de la muestra seleccionada, sin ponderación, y buscará reconocer distribuciones y patrones observables. Cada registro corresponde a una persona; no se trata de respuestas emparejadas de ambos integrantes de una pareja. Las diferencias observadas no se interpretarán como relaciones causales.

El volumen y la estructura de los datos requieren un procedimiento reproducible para seleccionar a las personas de interés, interpretar los códigos de respuesta, revisar valores faltantes y comprobar la consistencia del conjunto resultante. ## Pregunta de investigación

¿Cómo varían la valoración de la vida sexual y el bienestar mental o emocional percibido según el grado de dependencia económica de la pareja entre las personas encuestadas en ENSSEX 2022–2023 que conviven con ella?

El fenómeno analizado corresponde a las posibles diferencias en la valoración de la vida sexual y en el bienestar mental o emocional percibido entre personas que conviven con su pareja, considerando su grado de dependencia económica respecto de ella. Se compararán las respuestas entre quienes declaran dependencia completa, parcial o ausencia de dependencia, utilizando la información disponible en ENSSEX 2022–2023.

Las tres variables centrales son la dependencia económica de la pareja (`p93`), la valoración de la vida sexual (`i_6_p9`) y el bienestar mental o emocional percibido (`i_2_p9`). Las dos valoraciones se estudiarán por separado: aunque ambas usan una escala de 1 a 7, representan ámbitos distintos y no se sumarán en un único indicador. El bienestar percibido no se interpretará como un diagnóstico clínico.

## Objetivos del proyecto

### Objetivo general

Analizar cómo varían la valoración de la vida sexual y el bienestar mental o emocional percibido según el grado de dependencia económica de la pareja en personas encuestadas en ENSSEX 2022–2023 que conviven con ella.

### Objetivos específicos

1. Identificar a las personas encuestadas que declaran convivir con su pareja y conformar con ellas la población de análisis.
2. Describir la distribución del grado de dependencia económica de la pareja en la población seleccionada.
3. Caracterizar la valoración de la vida sexual y el bienestar mental o emocional percibido según los distintos grados de dependencia económica.
4. Comparar ambas dimensiones de bienestar entre los grupos definidos por nivel de dependencia económica, identificando diferencias y patrones relevantes.

En estos objetivos, «población de análisis» se refiere al subconjunto de personas encuestadas que cumple los criterios de inclusión. Los resultados describirán esta muestra sin aplicar factores de expansión, de acuerdo con el alcance del proyecto.

## Grupo de estudio y unidad de análisis

La unidad de análisis será cada persona encuestada. El grupo de interés estará formado por quienes declaran tener pareja y vivir actualmente con ella. Ambas condiciones deberán cumplirse conjuntamente:

| Criterio | Variable y pregunta | Respuesta de inclusión |
|---|---|---|
| Tener pareja actualmente | `p81`: «Actualmente, ¿usted tiene pareja?» | `1 = Sí`. |
| Convivir con esa pareja | `p83`: «¿Y actualmente usted vive con esa pareja?» | `1 = Sí`. |

El criterio será **`p81 = 1` y `p83 = 1`**. El estado civil no sustituirá estas preguntas y no se exigirá haber tenido relaciones sexuales con la pareja durante el último mes. Una respuesta de desconocimiento, rechazo o ausencia por salto del cuestionario no equivale a una respuesta negativa.

La revisión preliminar identificó 8.579 registros que cumplen los dos criterios, con folio presente y único. De ellos, 8.392 contienen respuestas sustantivas válidas simultáneamente en las tres variables centrales. Estos son antecedentes de disponibilidad que deberán reproducirse en F2, no una decisión de eliminar registros. El número de respuestas válidas podrá cambiar según las variables de cada comparación.

La encuesta está dirigida a personas de 18 años o más. La edad máxima de 100 años observada en esta copia es una característica de los datos y no un límite de inclusión definido por el proyecto. El filtro y la selección de columnas se implementarán en F2; en F1 se documentan las decisiones y se reconoce la base general.

### Diferencias entre las fuentes y criterio de documentación

**Salud general (`p10`).** El cuestionario, página 4, dice «3 = Regular»; los metadatos del SAV y el libro de códigos, página 14, dicen «3 = Ni buena ni mala». El diccionario conserva la etiqueta de la base y deja registrada la formulación del cuestionario. El código 3 sigue siendo válido; no se reemplaza por un faltante. Si posteriormente se elige una etiqueta de presentación, se indicará su procedencia sin modificar silenciosamente el dato original.

**Convivencia (`p83`).** El cuestionario, página 22, muestra «99 = NR»; la versión local del SAV y el libro de códigos, página 65, utilizan «9 = NS/NR». La comprobación del CSV completo encontró valores 1, 2 y 9, además de vacíos; no encontró 99. Para esta versión, el código operativo de no respuesta es **9**. El 99 se documenta como discrepancia de la fuente, no como valor observado. Si apareciera en otra versión, debe verificarse antes de aplicar las reglas.

El SAV marca como «scale» estas variables, incluso las categóricas y el identificador. Se utiliza la clasificación analítica de este diccionario según el significado de cada variable, en lugar de copiar esa marca automáticamente. Los metadatos no declaran rangos de faltantes de usuario (`missing_ranges` está vacío); por ello, los códigos especiales requieren tratamiento explícito por variable.

**Inicio de convivencia y edad de la pareja (`p84` y `p91`).** El cuestionario establece `9999 = NR` para P84 (p. 22) y `999 = NS/NR` para P91 (p. 23). Los metadatos del SAV no asignan etiquetas a estos códigos; el libro de códigos informa rangos de 1942–2022 y 18–100, respectivamente (pp. 65 y 67). Los códigos especiales se conservan en el diccionario como previsión documental y su frecuencia se comprueba en el reconocimiento. No se atribuye a otros valores, como 99 en una edad, el significado de no respuesta sin respaldo de la fuente.

**Género (`p3`).** Se conserva la pregunta original P3 y sus categorías, diferenciándola de `p1`, que mide sexo asignado al nacer. La base también contiene columnas denominadas `genero` y `genero_separada`, con clasificaciones distintas; esas recodificaciones no sustituyen a P3 en esta selección. No se incorpora el texto libre de «Otro» a los resultados.

**Referencia temporal (`fecha`).** El libro de códigos incluye esta columna, pero omite su descripción y contenido (p. 10), y los metadatos del SAV no aportan una etiqueta. Se conserva como auxiliar para revisar su significado y calidad antes de utilizarla como fecha de encuesta. El reconocimiento muestra fechas de 2022 y dos registros con `1/01/1970`; estos últimos se señalan para revisión. No se utiliza el año actual ni se supone que todas las entrevistas ocurrieron en 2023 para calcular una duración.

## Funciones para comprobar el entorno y la estructura

Separamos la revisión de bibliotecas y la revisión de carpetas en dos funciones. Cada una recibe las rutas que necesita y devuelve una tabla de evidencia. Una biblioteca ausente, una versión diferente o una carpeta requerida inexistente detiene el flujo con un mensaje que permite ubicar el problema.

Las versiones se consultan con [`importlib.metadata`](https://docs.python.org/3.13/library/importlib.metadata.html) y se contrastan con todas las dependencias fijadas en `requirements.txt`.

```python
def verificar_dependencias(ruta):
    """Contrasta las versiones instaladas con el archivo de dependencias."""
    registros = []
    for linea in ruta.read_text(encoding="utf-8-sig").splitlines():
        linea = linea.strip()
        if not linea or linea.startswith("#"):
            continue
        if "==" not in linea:
            raise ValueError(f"Dependencia sin versión exacta: {linea}")
        paquete, esperada = linea.split("==", 1)
        try:
            instalada = version(paquete)
        except PackageNotFoundError:
            instalada = "No instalada"
        registros.append({
            "Paquete": paquete, "Esperada": esperada, "Instalada": instalada,
            "Coincide": instalada == esperada,
        })
    if not registros:
        raise ValueError("El archivo de dependencias no contiene versiones verificables.")
    resultado = pd.DataFrame(registros)
    diferencias = resultado.loc[~resultado["Coincide"], "Paquete"].tolist()
    if diferencias:
        raise RuntimeError(
            "Revisar el entorno e instalar requirements.txt. Diferencias: "
            + ", ".join(diferencias)
        )
    return resultado


def verificar_estructura(raiz):
    """Comprueba las carpetas y documentos que necesita el flujo de F1."""
    carpetas = [
        "F1/notebooks", "data/raw", "F2/data/processed", "src", "F2/src", "F2/tests", "F1/docs",
        "F2/notebooks", "F2/docs",
    ]
    archivos = [
        "requirements.txt", "docs/datos/verificacion_conversion_enssex.json",
        "F1/docs/Instalacion_y_ejecucion.md", "docs/datos/Obtencion_y_conversion_ENSSEX.md",
    ]
    registros = [
        {"Ruta relativa": ruta, "Tipo": "Carpeta", "Existe": (raiz / ruta).is_dir()}
        for ruta in carpetas
    ] + [
        {"Ruta relativa": ruta, "Tipo": "Archivo", "Existe": (raiz / ruta).is_file()}
        for ruta in archivos
    ]
    resultado = pd.DataFrame(registros)
    ausentes = resultado.loc[~resultado["Existe"], "Ruta relativa"].tolist()
    if ausentes:
        raise FileNotFoundError("Faltan elementos del repositorio: " + ", ".join(ausentes))
    return resultado
```

## Lectura de la base y comprobación de su identidad

Antes de leer el CSV comprobamos que exista y que su huella SHA-256 coincida con la evidencia de conversión conservada en el repositorio. También revisamos la cabecera para detectar nombres repetidos antes de que el lector pueda renombrarlos automáticamente.

La lectura mantiene todas las filas y columnas. Usamos `keep_default_na=False` y `na_values=[""]`: los campos vacíos se representan como ausencias, pero un texto literal como `NA` se conserva. Los códigos numéricos de no respuesta permanecen sin recodificar. Estas opciones siguen la [documentación de `pandas.read_csv`](https://pandas.pydata.org/docs/reference/api/pandas.read_csv.html).

La huella de la tabla en memoria combina los valores, el índice, los nombres de las columnas y sus tipos. Se utiliza para comparar el estado inicial y final de esta ejecución, no como un identificador permanente entre versiones de pandas.

```python
def calcular_sha256(ruta):
    """Lee el archivo por bloques y calcula su huella de integridad."""
    huella = hashlib.sha256()
    with ruta.open("rb") as archivo:
        for bloque in iter(lambda: archivo.read(1024 * 1024), b""):
            huella.update(bloque)
    return huella.hexdigest()


def validar_columnas(tabla, requeridas):
    """Exige registros, nombres únicos y las columnas necesarias para el estudio."""
    if tabla.empty:
        raise ValueError("La base no contiene registros para reconocer.")
    if not tabla.columns.is_unique:
        raise ValueError("La base contiene nombres de columnas repetidos.")
    faltantes = sorted(set(requeridas) - set(tabla.columns))
    if faltantes:
        raise ValueError("Faltan variables del proyecto: " + ", ".join(faltantes))


def cargar_base(ruta, separador, codificacion, requeridas, forma_esperada, huella_esperada):
    """Carga íntegramente el CSV de la versión documentada y verifica su estructura."""
    if not ruta.is_file():
        raise FileNotFoundError(
            f"Falta {ruta}. Consultar la guía de obtención y conversión de ENSSEX."
        )
    if calcular_sha256(ruta) != huella_esperada:
        raise ValueError("El CSV no coincide con la versión registrada en la conversión.")
    with ruta.open(encoding=codificacion, newline="") as archivo:
        cabecera = next(csv.reader(archivo, delimiter=separador), [])
    if not cabecera or len(cabecera) != len(set(cabecera)):
        raise ValueError("La cabecera está vacía o contiene nombres repetidos.")
    if len(cabecera) != forma_esperada[1]:
        raise ValueError("Revisar el separador y la cantidad de columnas del CSV.")
    tabla = pd.read_csv(
        ruta, sep=separador, encoding=codificacion,
        keep_default_na=False, na_values=[""], low_memory=False,
    )
    validar_columnas(tabla, requeridas)
    if tabla.shape != tuple(forma_esperada):
        raise ValueError(f"Dimensiones inesperadas: {tabla.shape}; se esperaba {forma_esperada}.")
    if tabla.columns.tolist() != cabecera:
        raise ValueError("Los nombres o el orden de columnas cambiaron durante la lectura.")
    return tabla


def huella_tabla(tabla):
    """Obtiene una firma para detectar cambios durante esta misma ejecución."""
    huella = hashlib.sha256(pd.util.hash_pandas_object(tabla, index=True).to_numpy().tobytes())
    esquema = list(zip(tabla.columns.tolist(), tabla.dtypes.astype(str).tolist()))
    huella.update(json.dumps(esquema, ensure_ascii=False).encode("utf-8"))
    return huella.hexdigest()
```

## Reconocimiento de las variables acordadas

La siguiente función distingue tres aspectos: el tipo analítico definido por el equipo, el tipo asignado durante la lectura y la disponibilidad de respuestas. Los valores distintos incluyen los códigos especiales, mientras que los vacíos y la no respuesta codificada se cuentan por separado.

El resumen utiliza todos los registros de ENSSEX. Por eso, los vacíos de preguntas sobre la pareja pueden corresponder a saltos del cuestionario. El diagnóstico detallado de esos saltos y la preparación del grupo de convivientes se desarrollarán en F2.

Para `fecha` se realiza además una lectura temporal de diagnóstico en una serie independiente, conservando la columna original. Se cuentan valores no interpretables y fechas fuera del período nominal 2022–2023. Esta comprobación detecta anomalías para F2; no confirma por sí sola qué representa la fecha ni autoriza su uso en una variable derivada.

```python
def resumir_variables(tabla, especificaciones):
    """Describe las variables acordadas sin recodificar ni modificar la tabla."""
    validar_columnas(tabla, especificaciones)
    registros = []
    for nombre, (tipo_analitico, codigos_especiales) in especificaciones.items():
        serie = tabla[nombre]
        registros.append({
            "Variable": nombre,
            "Tipo analítico": tipo_analitico,
            "Tipo de lectura": str(serie.dtype),
            "Valores distintos": int(serie.nunique(dropna=True)),
            "Vacíos": int(serie.isna().sum()),
            "No respuesta codificada": int(serie.isin(codigos_especiales).sum()),
        })
    return pd.DataFrame(registros)


perfil_variables = resumir_variables(datos, VARIABLES)
display(perfil_variables)

# Diagnóstico temporal en una serie separada; no modifica datos["fecha"].
fechas_revision = pd.to_datetime(datos["fecha"], format="%d/%m/%Y", errors="coerce")
fuera_periodo = fechas_revision.notna() & ~fechas_revision.dt.year.isin([2022, 2023])
diagnostico_fecha = pd.DataFrame({
    "Comprobación temporal": [
        "Fecha ausente en el archivo",
        "Texto presente que no se puede interpretar como día/mes/año",
        "Fechas interpretables fuera de 2022–2023",
    ],
    "Registros": [
        int(datos["fecha"].isna().sum()),
        int((datos["fecha"].notna() & fechas_revision.isna()).sum()),
        int(fuera_periodo.sum()),
    ],
})
display(diagnostico_fecha)
```
