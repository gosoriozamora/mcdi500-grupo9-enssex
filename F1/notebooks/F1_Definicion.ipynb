## Presentación y contextualización

La Encuesta Nacional de Salud, Sexualidad y Género (ENSSEX) 2022–2023, del Ministerio de Salud de Chile, reúne información sobre salud, sexualidad, bienestar y relaciones de pareja en personas adultas. Elegimos esta fuente porque permite identificar a quienes tienen pareja y conviven con ella, y conocer tanto su dependencia económica como sus valoraciones sobre la vida sexual y el bienestar mental o emocional (Ministerio de Salud, s. f.; Ministerio de Salud, 2022).

Nuestro interés es reconocer si esas dos valoraciones presentan diferencias entre quienes dependen completamente de su pareja, quienes dependen en parte y quienes declaran no depender económicamente de ella. La convivencia ofrece un contexto compartido para comparar estos grupos, pero no convierte a las personas encuestadas en parejas emparejadas para el análisis ni permite establecer relaciones causales.

La utilidad del proyecto está en describir estas diferencias de forma clara y reproducible, considerando las características de las personas y las limitaciones de las respuestas disponibles. Conservaremos la valoración de la vida sexual y el bienestar mental o emocional como dos resultados distintos. Este último recoge una percepción personal y no se interpretará como un diagnóstico clínico.

El volumen y la estructura de la base hacen necesario utilizar programación. Necesitamos comprobar nombres de columnas, interpretar códigos de respuesta, aplicar criterios de selección consistentes y registrar cómo cambia el conjunto de trabajo durante su preparación. ## Pregunta de investigación

¿Cómo varían la valoración de la vida sexual y el bienestar mental o emocional percibido según el grado de dependencia económica de la pareja entre las personas encuestadas en ENSSEX 2022–2023 que conviven con ella?

La pregunta relaciona tres variables centrales: dependencia económica de la pareja (`p93`), valoración de la vida sexual (`i_6_p9`) y bienestar mental o emocional percibido (`i_2_p9`). Su propósito es describir y comparar las respuestas entre grupos; no demostrar que la dependencia económica provoque cambios en el bienestar.

## Objetivos del proyecto

### Objetivo general

Analizar cómo varían la valoración de la vida sexual y el bienestar mental o emocional percibido según el grado de dependencia económica de la pareja en personas encuestadas en ENSSEX 2022–2023 que conviven con ella.

### Objetivos específicos

1. Identificar a las personas encuestadas que declaran convivir con su pareja y conformar con ellas la población de análisis.
2. Describir la distribución del grado de dependencia económica de la pareja en la población seleccionada.
3. Caracterizar la valoración de la vida sexual y el bienestar mental o emocional percibido según los distintos grados de dependencia económica.
4. Comparar ambas dimensiones de bienestar entre los grupos definidos por nivel de dependencia económica, identificando diferencias y patrones relevantes.

En estos objetivos, «población de análisis» se refiere al subconjunto de personas encuestadas que cumple los criterios de inclusión. Los resultados describirán esta muestra sin aplicar factores de expansión, de acuerdo con el alcance del proyecto.

### Ponderación y alcance de las conclusiones

Para este proyecto describiremos los registros seleccionados sin aplicar factores de expansión: cada respuesta válida tendrá el mismo peso en los conteos y porcentajes. Esta decisión es coherente con una pregunta centrada en las personas encuestadas y con el trabajo de preparación y comparación descriptiva previsto.

El Manual de uso de base de datos y libro de códigos de ENSSEX señala que, sin factores de expansión, los resultados describen la muestra particular recolectada (Ministerio de Salud, 2022, pp. 7–9). Por eso, nuestras conclusiones se expresarán como diferencias observadas **entre las personas encuestadas incluidas en el análisis**. No se presentarán como estimaciones de todas las personas que conviven en Chile ni como resultados representativos de cada región. La magnitud y dirección de las diferencias podrían cambiar con un análisis ponderado.

La respuesta a la pregunta se apoyará en frecuencias, porcentajes por categoría y resúmenes apropiados para variables ordinales. Cada tabla o gráfico indicará cuántas respuestas se utilizaron y cómo se trataron las ausencias. Las comparaciones serán descriptivas; no se plantean pruebas de significación ni intervalos de confianza poblacionales dentro de este alcance.

Si más adelante se amplía el propósito a inferencia poblacional, será necesario revisar la selección de variables e incorporar el ponderador y el diseño muestral correspondientes, según el manual de ENSSEX. Esa ampliación deberá documentarse antes de aplicarse y no forma parte de la selección actual de 15 columnas.

## Funciones para comprobar el entorno y la estructura

Separamos la revisión de bibliotecas y la revisión de carpetas en dos funciones. Cada una recibe las rutas que necesita y devuelve una tabla de evidencia. Una biblioteca ausente, una versión diferente o una carpeta requerida inexistente detiene el flujo con un mensaje que permite ubicar el problema.

Las versiones se consultan con [`importlib.metadata`](https://docs.python.org/3.13/library/importlib.metadata.html) y se contrastan con todas las dependencias fijadas en `requirements.txt`.

```python
def verificar_dependencias(ruta):
    """Contrasta las versiones instaladas con el archivo de dependencias."""
    registros = []
    for linea in ruta.read_text(encoding="utf-8-sig").splitlines():
        linea = linea.strip()
        if not linea or linea.startswith("#"):
            continue
        if "==" not in linea:
            raise ValueError(f"Dependencia sin versión exacta: {linea}")
        paquete, esperada = linea.split("==", 1)
        try:
            instalada = version(paquete)
        except PackageNotFoundError:
            instalada = "No instalada"
        registros.append({
            "Paquete": paquete, "Esperada": esperada, "Instalada": instalada,
            "Coincide": instalada == esperada,
        })
    if not registros:
        raise ValueError("El archivo de dependencias no contiene versiones verificables.")
    resultado = pd.DataFrame(registros)
    diferencias = resultado.loc[~resultado["Coincide"], "Paquete"].tolist()
    if diferencias:
        raise RuntimeError(
            "Revisar el entorno e instalar requirements.txt. Diferencias: "
            + ", ".join(diferencias)
        )
    return resultado


def verificar_estructura(raiz):
    """Comprueba las carpetas y documentos que necesita el flujo de F1."""
    carpetas = [
        "F1/notebooks", "F1/data/raw", "F1/data/processed", "F1/src", "F1/docs",
        "F2/notebooks", "F2/docs",
    ]
    archivos = [
        "requirements.txt", "F1/docs/verificacion_conversion_enssex.json",
        "F1/docs/Instalacion_y_ejecucion.md", "F1/docs/Obtencion_y_conversion_ENSSEX.md",
    ]
    registros = [
        {"Ruta relativa": ruta, "Tipo": "Carpeta", "Existe": (raiz / ruta).is_dir()}
        for ruta in carpetas
    ] + [
        {"Ruta relativa": ruta, "Tipo": "Archivo", "Existe": (raiz / ruta).is_file()}
        for ruta in archivos
    ]
    resultado = pd.DataFrame(registros)
    ausentes = resultado.loc[~resultado["Existe"], "Ruta relativa"].tolist()
    if ausentes:
        raise FileNotFoundError("Faltan elementos del repositorio: " + ", ".join(ausentes))
    return resultado
```

## Lectura de la base y comprobación de su identidad

Antes de leer el CSV comprobamos que exista y que su huella SHA-256 coincida con la evidencia de conversión conservada en el repositorio. También revisamos la cabecera para detectar nombres repetidos antes de que el lector pueda renombrarlos automáticamente.

La lectura mantiene todas las filas y columnas. Usamos `keep_default_na=False` y `na_values=[""]`: los campos vacíos se representan como ausencias, pero un texto literal como `NA` se conserva. Los códigos numéricos de no respuesta permanecen sin recodificar. Estas opciones siguen la [documentación de `pandas.read_csv`](https://pandas.pydata.org/docs/reference/api/pandas.read_csv.html).

La huella de la tabla en memoria combina los valores, el índice, los nombres de las columnas y sus tipos. Se utiliza para comparar el estado inicial y final de esta ejecución, no como un identificador permanente entre versiones de pandas.

```python
def calcular_sha256(ruta):
    """Lee el archivo por bloques y calcula su huella de integridad."""
    huella = hashlib.sha256()
    with ruta.open("rb") as archivo:
        for bloque in iter(lambda: archivo.read(1024 * 1024), b""):
            huella.update(bloque)
    return huella.hexdigest()


def validar_columnas(tabla, requeridas):
    """Exige registros, nombres únicos y las columnas necesarias para el estudio."""
    if tabla.empty:
        raise ValueError("La base no contiene registros para reconocer.")
    if not tabla.columns.is_unique:
        raise ValueError("La base contiene nombres de columnas repetidos.")
    faltantes = sorted(set(requeridas) - set(tabla.columns))
    if faltantes:
        raise ValueError("Faltan variables del proyecto: " + ", ".join(faltantes))


def cargar_base(ruta, separador, codificacion, requeridas, forma_esperada, huella_esperada):
    """Carga íntegramente el CSV de la versión documentada y verifica su estructura."""
    if not ruta.is_file():
        raise FileNotFoundError(
            f"Falta {ruta}. Consultar la guía de obtención y conversión de ENSSEX."
        )
    if calcular_sha256(ruta) != huella_esperada:
        raise ValueError("El CSV no coincide con la versión registrada en la conversión.")
    with ruta.open(encoding=codificacion, newline="") as archivo:
        cabecera = next(csv.reader(archivo, delimiter=separador), [])
    if not cabecera or len(cabecera) != len(set(cabecera)):
        raise ValueError("La cabecera está vacía o contiene nombres repetidos.")
    if len(cabecera) != forma_esperada[1]:
        raise ValueError("Revisar el separador y la cantidad de columnas del CSV.")
    tabla = pd.read_csv(
        ruta, sep=separador, encoding=codificacion,
        keep_default_na=False, na_values=[""], low_memory=False,
    )
    validar_columnas(tabla, requeridas)
    if tabla.shape != tuple(forma_esperada):
        raise ValueError(f"Dimensiones inesperadas: {tabla.shape}; se esperaba {forma_esperada}.")
    if tabla.columns.tolist() != cabecera:
        raise ValueError("Los nombres o el orden de columnas cambiaron durante la lectura.")
    return tabla


def huella_tabla(tabla):
    """Obtiene una firma para detectar cambios durante esta misma ejecución."""
    huella = hashlib.sha256(pd.util.hash_pandas_object(tabla, index=True).to_numpy().tobytes())
    esquema = list(zip(tabla.columns.tolist(), tabla.dtypes.astype(str).tolist()))
    huella.update(json.dumps(esquema, ensure_ascii=False).encode("utf-8"))
    return huella.hexdigest()
```

### Carga completa y resumen general

Las dimensiones esperadas proceden de la evidencia de conversión. Mostramos el tamaño del archivo, los tipos de lectura y la presencia de identificadores o filas repetidas. Estos resultados describen la base completa; no se aplica todavía el filtro de convivencia ni se eliminan registros.

```python
conversion = json.loads(RUTA_CONVERSION.read_text(encoding="utf-8"))
assert conversion["status"] == "OK", "Revisar la evidencia de conversión."
assert conversion["csv_file"] == RUTA_DATOS.name, "La evidencia corresponde a otro archivo."
forma_esperada = (conversion["checks"]["rows"], conversion["checks"]["columns"])

datos = cargar_base(
    RUTA_DATOS, SEPARADOR, CODIFICACION, VARIABLES,
    forma_esperada, conversion["csv_sha256"],
)
firma_inicial = huella_tabla(datos)
filas_duplicadas = int(datos.duplicated().sum())
folios_ausentes = int(datos["folio_encuesta"].isna().sum())
folios_repetidos = int(datos["folio_encuesta"].dropna().duplicated().sum())

resumen_general = pd.DataFrame({
    "Indicador": [
        "Filas", "Columnas", "Variables del proyecto presentes", "Tamaño del CSV (MiB)",
        "Filas repetidas exactas", "Folios ausentes", "Folios repetidos adicionales",
    ],
    "Resultado": [
        str(len(datos)), str(len(datos.columns)), str(len(VARIABLES)),
        f"{RUTA_DATOS.stat().st_size / 1024**2:.2f}",
        str(filas_duplicadas), str(folios_ausentes), str(folios_repetidos),
    ],
})
display(resumen_general)
display(
    datos.dtypes.astype(str).value_counts().rename_axis("Tipo de lectura")
    .reset_index(name="Cantidad de columnas")
)
print("SHA-256 del CSV verificado:", conversion["csv_sha256"])
```

## Reconocimiento de las variables acordadas

La siguiente función distingue tres aspectos: el tipo analítico definido por el equipo, el tipo asignado durante la lectura y la disponibilidad de respuestas. Los valores distintos incluyen los códigos especiales, mientras que los vacíos y la no respuesta codificada se cuentan por separado.

```python
def resumir_variables(tabla, especificaciones):
    """Describe las variables acordadas sin recodificar ni modificar la tabla."""
    validar_columnas(tabla, especificaciones)
    registros = []
    for nombre, (tipo_analitico, codigos_especiales) in especificaciones.items():
        serie = tabla[nombre]
        registros.append({
            "Variable": nombre,
            "Tipo analítico": tipo_analitico,
            "Tipo de lectura": str(serie.dtype),
            "Valores distintos": int(serie.nunique(dropna=True)),
            "Vacíos": int(serie.isna().sum()),
            "No respuesta codificada": int(serie.isin(codigos_especiales).sum()),
        })
    return pd.DataFrame(registros)


perfil_variables = resumir_variables(datos, VARIABLES)
display(perfil_variables)
```
